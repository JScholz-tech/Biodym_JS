```python
import pandas as pd
import numpy as np

# Load both Excel files
file1 = "Case_study_results.xlsx"
file2 = "byhand.xlsx"

xls1 = pd.ExcelFile(file1)
xls2 = pd.ExcelFile(file2)

total_difference = 0  # To store the sum of all differences across sheets

# Ensure both have the same sheets
if set(xls1.sheet_names) != set(xls2.sheet_names):
    print("Sheet names do not match!")
else:
    for sheet in xls1.sheet_names:
        df1 = pd.read_excel(xls1, sheet_name=sheet)
        df2 = pd.read_excel(xls2, sheet_name=sheet)
        
        # Ensure both DataFrames have the same shape
        if df1.shape != df2.shape:
            print(f"Sheet '{sheet}' has different dimensions!")
            continue
        
        # Convert non-numeric values to NaN to prevent errors
        df1_numeric = df1.apply(pd.to_numeric, errors='coerce')
        df2_numeric = df2.apply(pd.to_numeric, errors='coerce')

        # Subtract the two DataFrames
        diff = df1_numeric - df2_numeric
        
        # Sum absolute differences for the sheet
        sheet_difference = diff.abs().sum().sum()
        total_difference += sheet_difference

        print(f"Total difference in sheet '{sheet}': {sheet_difference}")

# Print the grand total difference across all sheets
print(f"Grand total difference across all sheets: {total_difference}")
```
