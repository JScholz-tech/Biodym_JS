```python
import numpy as np
import pandas as pd

# --- 1. Data Collection and Preparation ---
# Hypothetical historical annual rye straw harvest data (in tonnes) for 2010-2020
years_historical = np.arange(2010, 2025)
# Example: Grain harvest (and a straw-to-grain ratio is used to get straw)
# Or, this is directly the straw harvest data.
historical_harvest_data = np.array([ 
    783485, 589600, 974100, 1188700, 954600, 863400, 807600, 614800, 497000, 716100, 827800, 613200, 578600, 570000, 472200
])



print(f"Historical Data (2010-2024):\n{historical_harvest_data}\n")

# --- 2. Calculate the Median ---
median_harvest = np.median(historical_harvest_data)
print(f"Median of Historical Harvest: {median_harvest:.2f} (tonnes)\n")

# --- 3. Calculate the Standard Deviation ---
# Using sample standard deviation (ddof=1)
std_dev_harvest = np.std(historical_harvest_data, ddof=1)
print(f"Standard Deviation of Historical Harvest: {std_dev_harvest:.2f} (tonnes)\n")

# --- 4. Generate Forecasted Values (2025-2050) ---
forecast_years = np.arange(2025, 2051)
forecasted_harvest_values = []

# Seed for reproducibility (optional)
np.random.seed(42)

for year in forecast_years:
    # Using a random value from a Standard Normal Distribution N(0,1)
    random_factor = np.random.normal(0, 1)
    forecast_value = median_harvest + (random_factor * std_dev_harvest)
    
    # Ensure non-negativity (important if std_dev is large compared to median)
    forecast_value = max(0, forecast_value)
    forecasted_harvest_values.append(forecast_value)

forecast_df = pd.DataFrame({
    'Year': forecast_years,
    'Forecasted Rye Harvest (tonnes)': forecasted_harvest_values
})

print("Forecasted Values (2025-2050):")
print(forecast_df)

# --- Optional: Consider a simpler range for scenarios ---
forecast_low_scenario = median_harvest - std_dev_harvest
forecast_high_scenario = median_harvest + std_dev_harvest
print(f"\nSimple Scenario Range based on one StdDev:")
print(f"Low Forecast (Median - 1*StdDev): {max(0, forecast_low_scenario):.2f}")
print(f"Median Forecast: {median_harvest:.2f}")
print(f"High Forecast (Median + 1*StdDev): {forecast_high_scenario:.2f}")
```
